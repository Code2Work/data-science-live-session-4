## 🟠 Bölüm 1 — Eski Yöntem: `%` Formatting

1990'lardan kalma, hala bazı kodlarda görürsünüz. C dilinden gelme.

````python
# Tek değer
name = "Ali"
print("Merhaba %s" % name)

# Birden fazla değer → TUPLE içinde ver
age = 25
print("Merhaba %s, yaşın %d" % (name, age))
````

````python
# Format kodları:
# %s = string
# %d = integer (sayı)
# %f = float (ondalık)
# %.2f = float, 2 ondalık

salary = 12500.5
print("Maaş: %.2f yasin: %d ismin: %s" % (salary, age, name))        # → 12500.50
print("Maaş: %d ₺" % salary)         # → 12500 (int'e çevirdi)
````

**Problemleri:**
- Tip kodları (`%s`, `%d`) ezberlemek lazım
- Birden fazla değer için **tuple** gerekiyor
- Yanlış tip → hata (`%d` ile string verirsen patlar)

> Bu yöntem **eskiyor**. Görürseniz tanıyın, yazmaya gerek yok.

---
## 🟡 Bölüm 2 — Orta Yöntem: `.format()`

Python 2.6 ile geldi. `%`'dan daha esnek.

````python
name = "Ali"
age = 25

# Sıralı (boş süslü parantez)
print("Merhaba {}, yaşın {}".format(name, age))

# İsimli (anahtar kelime)
print("Merhaba {ad}, yaşın {yas}".format(ad=name, yas=age))

# İndeksli
print("{0} {1} {0}".format("Merhaba", "Ali"))  # → Merhaba Ali Merhaba
````

**`.format()` daha iyi ama hala bir sorun var:**
Değişkenler **stringin dışında**. Okuduğunda gözün ileri-geri gitmek zorunda.

```python
# Hangi {} hangi değişken? Karıştırması kolay
print("{} {} {} {}".format(a, b, c, d))
```

## 🟢 Bölüm 3 — Modern Yöntem: f-string ⭐

Python 3.6+ ile geldi. **Şu an en popüler ve önerilen** yöntem.

**f**ormatted string — başına `f` koyarsın, içinde `{değişken}` yazarsın.

````python
name = "Ali"
age = 25

print(f"Merhaba {name}, yaşın {age}")

# Değişkenler stringin İÇİNDE — okumak çok kolay
````

````python
# İçinde HESAPLAMA bile yapabilirsin
print(f"5 yıl sonra {age + 5} yaşında olacak")
print(f"İsmin uzunluğu: {len(name)} karakter")
print(f"Büyük harfle: {name.upper()}")
````

> **Mesaj:** f-string içinde **Python ifadesi** yazabilirsin. Değişken, fonksiyon, hesap — ne istersen.

### Karşılaştırma — 3 yöntem yan yana

````python
name = "Ali"
age = 25

# 🔴 Eski
print("Merhaba %s, yaşın %d" % (name, age))

# 🟡 Orta
print("Merhaba {}, yaşın {}".format(name, age))

# 🟢 Modern ⭐
print(f"Merhaba {name}, yaşın {age}")
````

## 🎯 Bölüm 4 — Format Spec'leri (f-string'in süper gücü)

`{değişken:format}` syntax'ı ile sayıları, hizalamayı, kesirleri kontrol ediyorsun.

### 4a · Ondalık basamak

````python
import math

pi = math.pi  # 3.141592653589793

print(f"Pi = {pi}")          # tam değer
print(f"Pi = {pi:.2f}")      # 2 ondalık → 3.14
print(f"Pi = {pi:.4f}")      # 4 ondalık → 3.1416
print(f"Pi = {pi:.0f}")      # 0 ondalık → 3
````

> `{değer:.2f}` → "2 ondalık basamaklı float" demek. **Veri analizinde her gün lazım.**

### 4b · Binlik ayraç

````python
nufus = 84680000
satis = 1250000.75

print(f"Türkiye nüfusu: {nufus}")        # 84680000 — okumak zor
print(f"Türkiye nüfusu: {nufus:,}")      # 84,680,000 — şık!

print(f"Toplam satış: {satis:,.1f} ₺")   # 1,250,000.75 ₺
````

> `{değer:,}` → virgüllü binlik ayraç
>
> `{değer:,.2f}` → hem virgül hem 2 ondalık

### 4c · Yüzde

````python
basari_orani = 0.8542

print(f"Başarı: {basari_orani}")          # 0.8542
print(f"Başarı: %{basari_orani * 100:.1f}") # %85.4
print(f"Başarı: {basari_orani:.1%}")      # 85.4% (otomatik *100)
````

> `{değer:.1%}` → yüzdeye çevirip 1 ondalık. Çok kullanışlı.

### 4d · Hizalama

````python
isimler = ["Ali", "Ayşe", "Mehmet", "Beyza"]
puanlar = [85, 92, 78, 96]

print("--- Hizasız ---")
for isim, puan in zip(isimler, puanlar):
    print(f"{isim}: {puan}")

print()
print("--- Hizalı ---")
for isim, puan in zip(isimler, puanlar):
    print(f"{isim:<10} {puan:>3}")  # sol 10, sağ 3 karakter
````

> `{değer:<10}` → 10 karakter, sola yasla
>
> `{değer:>10}` → 10 karakter, sağa yasla
>
> `{değer:^10}` → 10 karakter, ortala

### 4e · Dolgu karakteri

````python
for i in [1, 25, 142, 1500]:
    print(f"{i:>6}")    # boşlukla sağa yasla

print("---")

for i in [1, 25, 142, 1500]:
    print(f"{i:0>6}")   # SIFIR ile doldur
````

> `{değer:0>6}` → "6 karakter olacak şekilde **0** ile doldur, sağa yasla"
>
> Sipariş numarası, fatura ID, vb. için kullanışlı.

## 🎁 Bölüm 5 — Gerçek Örnek: Eurovision Raporu

````python
import psycopg2

conn = psycopg2.connect(
    host="localhost",
    database="postgres",
    user="postgres",
    password="1234",
)
cur = conn.cursor()
````

````python
# Türkiye'nin Eurovision istatistiklerini çek
cur.execute("""
    SELECT COUNT(*) AS katilim,
           AVG(points_final) AS ort,
           MAX(points_final) AS en_yuksek,
           MIN(points_final) AS en_dusuk
    FROM eurovision.contestants
    WHERE to_country = 'Turkey' AND points_final IS NOT NULL;
""")
katilim, ort, max_p, min_p = cur.fetchone()

# 🔴 Çirkin rapor
print("Turkey:", katilim, ort, max_p, min_p)
````

````python
# 🟢 Şık rapor — f-string ile
print("═══════════════════════════════════════════")
print("  Türkiye'nin Eurovision İstatistikleri")
print("═══════════════════════════════════════════")
print(f"  Katılım:     {katilim:>5} yarışma")
print(f"  Ortalama:    {ort:>5.1f} puan")
print(f"  En yüksek:   {max_p:>5.0f} puan")
print(f"  En düşük:    {min_p:>5.0f} puan")
print("═══════════════════════════════════════════")
````

````python
# Top 5 kazanan ülke — tablo gibi
cur.execute("""
    SELECT to_country, COUNT(*) AS kazanma
    FROM eurovision.contestants
    WHERE place_final = 1
    GROUP BY to_country
    ORDER BY kazanma DESC
    LIMIT 5;
""")
kazananlar = cur.fetchall()

print(f"{'ÜLKE':<20} {'KAZANMA':>10}")
print("─" * 32)
for ulke, sayi in kazananlar:
    print(f"{ulke:<20} {sayi:>10}")
````

````python
conn.close()
````

## 🎯 Kapanış

### Bugün ne öğrendik?

| Yöntem | Nasıl | Ne zaman |
|--------|-------|----------|
| `%` | `"%s, %d" % (a, b)` | 🚫 Eski, yazmaya gerek yok |
| `.format()` | `"{}, {}".format(a, b)` | 🟡 Eski kodda görürseniz |
| **f-string** | `f"{a}, {b}"` | ✅ Her yeni kodda kullan |

### Format spec özet

| Yazım | Anlamı |
|-------|--------|
| `{x:.2f}` | 2 ondalık |
| `{x:,}` | Binlik ayraç |
| `{x:,.2f}` | Hem binlik hem ondalık |
| `{x:.1%}` | Yüzde (×100) |
| `{x:>10}` | Sağa yasla, 10 karakter |
| `{x:<10}` | Sola yasla, 10 karakter |
| `{x:0>6}` | 6 karakter, 0 ile doldur |

### Ev ödevi

- 🟢 **Yeni başlayan**: Kendi ismini, yaşını, şehrini f-string ile yazdır. Sonra `.format()` ile aynısını yaz.
- 🟠 **Orta**: Eurovision'dan bir ülke seç, **Bölüm 5'teki gibi şık rapor** yaz.
- 🟣 **İleri**: Tablo formatında **en iyi 10 ülkeyi** kazanma sayısı + yüzde olarak göster (her ülke kazanma_sayısı / toplam_yarışma × 100).
